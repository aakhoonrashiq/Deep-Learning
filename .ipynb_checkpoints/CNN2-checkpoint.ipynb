# Image Classification on CIFAR-10
In this notebook, we will build a CNN from scratch to classify the CIFAR-10 dataset, and then compare it with a transfer learning approach using a pretrained ResNet-18 model.

## Limitations of CNNs
While Convolutional Neural Networks (CNNs) are extremely powerful for grid-like data (such as images), they have several limitations:
1. **Fixed-Size Grids**: CNNs generally expect fixed-size inputs. Although fully convolutional networks can handle varying resolutions, standard CNNs with fully connected layers at the end require a fixed image size.
2. **Variable-Length Sequential Data**: They cannot easily handle variable-length sequential data, such as sentences, text, or time-series.
3. **Long-Range Dependencies**: CNNs capture local features using small receptive fields. Capturing long-range dependencies requires many deep layers to expand the receptive field, which can be inefficient compared to other architectures.

These limitations motivate the use of Recurrent Neural Networks (RNNs) for sequential data, and more recently, Transformers, which use self-attention to effectively model long-range dependencies across the entire input sequence without the constraints of local receptive fields.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
from torchvision.models import resnet18, ResNet18_Weights

print("PyTorch version:", torch.__version__)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu" if torch.backends.mps.is_available() else "cpu")
if torch.backends.mps.is_available():
    device = torch.device("mps")
print("Using device:", device)
```

```python
class CIFAR10Kaggle(Dataset):
    def __init__(self, csv_file, root_dir, transform=None):
        self.labels_df = pd.read_csv(csv_file)
        self.root_dir = root_dir
        self.transform = transform
        
        # Map string labels to integers
        self.classes = sorted(self.labels_df['label'].unique())
        self.class_to_idx = {c: i for i, c in enumerate(self.classes)}
        
    def __len__(self):
        return len(self.labels_df)
    
    def __getitem__(self, idx):
        img_id = self.labels_df.iloc[idx, 0]
        label_str = self.labels_df.iloc[idx, 1]
        
        img_name = os.path.join(self.root_dir, f"{img_id}.png")
        image = Image.open(img_name).convert('RGB')
        
        label = self.class_to_idx[label_str]
        
        if self.transform:
            image = self.transform(image)
            
        return image, label

# Transforms
transform_train = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(32, padding=4),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2470, 0.2435, 0.2616)),
])

transform_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2470, 0.2435, 0.2616)),
])

csv_path = "trainLabels.csv"
img_dir = "cifar-10/train/"

# Create separate datasets for train and val to avoid transform overwriting
full_train_dataset = CIFAR10Kaggle(csv_file=csv_path, root_dir=img_dir, transform=transform_train)
full_val_dataset = CIFAR10Kaggle(csv_file=csv_path, root_dir=img_dir, transform=transform_test)

# Random split
indices = torch.randperm(len(full_train_dataset), generator=torch.Generator().manual_seed(42)).tolist()
train_size = int(0.8 * len(full_train_dataset))

train_dataset = Subset(full_train_dataset, indices[:train_size])
val_dataset = Subset(full_val_dataset, indices[train_size:])

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=128, shuffle=False, num_workers=2)

print(f"Train size: {len(train_dataset)}, Val size: {len(val_dataset)}")
```

```python
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super(SimpleCNN, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2, 2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )
        
    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

model_scratch = SimpleCNN(num_classes=10).to(device)
criterion = nn.CrossEntropyLoss()
optimizer_scratch = optim.Adam(model_scratch.parameters(), lr=0.001)

print(model_scratch)
```

```python
def train_model(model, train_loader, val_loader, criterion, optimizer, epochs=5):
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    
    for epoch in range(epochs):
        model.train()
        running_loss, correct, total = 0.0, 0, 0
        
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item() * inputs.size(0)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
            
        train_loss = running_loss / total
        train_acc = correct / total
        
        model.eval()
        val_loss, correct, total = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                
                val_loss += loss.item() * inputs.size(0)
                _, predicted = outputs.max(1)
                total += labels.size(0)
                correct += predicted.eq(labels).sum().item()
                
        val_loss /= total
        val_acc = correct / total
        
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        
        print(f"Epoch {epoch+1}/{epochs} - Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | Val Loss: {val_loss:.4f} Acc: {val_acc:.4f}")
        
    return history

print("Training CNN from scratch...")
history_scratch = train_model(model_scratch, train_loader, val_loader, criterion, optimizer_scratch, epochs=5)
```

```python
print("Setting up ResNet-18 for Transfer Learning...")
weights = ResNet18_Weights.IMAGENET1K_V1
model_transfer = resnet18(weights=weights)

# Freeze parameters
for param in model_transfer.parameters():
    param.requires_grad = False

# Replace the final fully connected layer for CIFAR-10 (10 classes)
num_ftrs = model_transfer.fc.in_features
model_transfer.fc = nn.Linear(num_ftrs, 10)
model_transfer = model_transfer.to(device)

optimizer_transfer = optim.Adam(model_transfer.fc.parameters(), lr=0.001)

print("Training Transfer Learning Model...")
# Let's use fewer epochs for ResNet to be faster, e.g. 3 epochs
history_transfer = train_model(model_transfer, train_loader, val_loader, criterion, optimizer_transfer, epochs=3)
```

```python
# Compare Validation Accuracies
plt.figure(figsize=(10, 5))
plt.plot(history_scratch['val_acc'], label='Scratch CNN Val Acc', marker='o')
plt.plot(history_transfer['val_acc'], label='ResNet-18 Transfer Val Acc', marker='s')
plt.title('Validation Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

print(f"Final Scratch CNN Accuracy: {history_scratch['val_acc'][-1]:.4f}")
print(f"Final ResNet-18 Transfer Accuracy: {history_transfer['val_acc'][-1]:.4f}")
```

```python
# Visualize activations of the first convolutional layer
def visualize_activations(model, img_tensor):
    model.eval()
    
    # Extract the first conv layer from our SimpleCNN
    conv_layer = model.features[0]
    
    with torch.no_grad():
        # Get feature maps
        activations = conv_layer(img_tensor.unsqueeze(0).to(device))
    
    activations = activations.squeeze(0).cpu() # [num_filters, H, W]
    num_filters = activations.shape[0]
    
    # Plot up to 32 filters
    fig, axes = plt.subplots(4, 8, figsize=(15, 8))
    fig.suptitle('Feature Maps of First Conv Layer (Simple CNN)', fontsize=16)
    
    for i, ax in enumerate(axes.flat):
        if i < num_filters:
            ax.imshow(activations[i], cmap='viridis')
            ax.axis('off')
    plt.tight_layout()
    plt.show()

# Get a sample image from validation dataset
sample_img, _ = val_dataset[0]

# Display original image
plt.figure(figsize=(3,3))
plt.title('Original Image')
# Unnormalize for visualization
mean = torch.tensor([0.4914, 0.4822, 0.4465]).view(3, 1, 1)
std = torch.tensor([0.2470, 0.2435, 0.2616]).view(3, 1, 1)
unnorm_img = sample_img * std + mean
unnorm_img = torch.clamp(unnorm_img, 0, 1)
plt.imshow(unnorm_img.permute(1, 2, 0).numpy())
plt.axis('off')
plt.show()

print("Visualizing learned filter activations...")
visualize_activations(model_scratch, sample_img)
```
